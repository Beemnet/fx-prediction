```python
import pandas as pd
import json
from urllib.parse import urlparse
from datetime import datetime
import os
```

#### google news

```python
# Read JSON data from google v22
with open('../../data/external/api_responses/google_news_v22_20250515_124215.json', 'r', encoding="utf-8") as f:
    data = json.load(f)
print(data)
```

```python
# Extract required fields
rows = []
items = data.get('data', [])
for item in items:
    # print(item)
    date_str = item.get('date', '')
    # date = datetime.strptime(date_str, '%Y-%m-%dT%HH:%MM:%SS.%fZ').strftime('%Y-%m-%d %H:%M:%S')
    date = datetime.fromisoformat(date_str.replace('Z', '+00:00')).strftime('%Y-%m-%d %H:%M:%S')
    url = item.get('url', '')
    url_no_http = url.replace('http://', '').replace('https://', '')
    domain = urlparse(url).netloc
    rows.append({'date': date, 'domain': domain, 'url': url_no_http})

df = pd.DataFrame(rows)
df.to_csv('google_news_v22_20250515_124215.csv', index=False)
```

```python
for file in ["google_news_v13_business_20250515_124213", "google_news_v13_technology_20250515_124213"]:
    with open(os.path.join('../../data/external/api_responses/', file+'.json'), 'r', encoding="utf-8") as f:
        data = json.load(f)
    print(data)
    # Extract required fields
    rows = []
    items = data.get('items', [])
    for item in items:
        # print(item)
        timestamp = item.get('timestamp', '')
        if timestamp:
            timestamp = int(timestamp) / 1000
            date = datetime.fromtimestamp((timestamp)).strftime('%Y-%m-%d %H:%M:%S')
        else: date  = ''
        url = item.get('newsUrl', '')
        url_no_http = url.replace('http://', '').replace('https://', '')
        domain = urlparse(url).netloc
        rows.append({'date': date, 'domain': domain, 'url': url_no_http})

    df = pd.DataFrame(rows)
    df.to_csv(f'{file}.csv', index=False)
```

#### real time news

```python
with open('../../data/external/api_responses/realtime_news_full_story_20250515_124224.json','r', encoding="utf-8") as f:
    data = json.load(f)
print(data)
```

```python
rows = []
items = data.get('data', []).get('all_articles', [])
for item in items:
    # print(item)
    date_str = item.get('published_datetime_utc', '')
    date = datetime.fromisoformat(date_str.replace('Z', '+00:00')).strftime('%Y-%m-%d %H:%M:%S')
    url = item.get('link', '')
    url_no_http = url.replace('http://', '').replace('https://', '')
    domain = urlparse(url).netloc
    rows.append({'date': date, 'domain': domain, 'url': url_no_http})

df = pd.DataFrame(rows)
df.to_csv('realtime_news_full_story_20250515_124224.csv', index=False)
```
